```python
import matplotlib.pyplot as plt

proportions = train[features].eq(1).mean().sort_values(ascending=False) * 100

plt.figure(figsize=(12, 6))
plt.bar(proportions.index, proportions.values)

plt.xlabel("Finding")
plt.ylabel("Positive Proportion (%)")
plt.title("Proportion of Positive Cases for Each Chest X-ray Finding")
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```
